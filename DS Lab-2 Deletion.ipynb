```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class CreateList:
    def __init__(self):
        self.head = None
        self.tail = None

    def add(self, data):
        newNode = Node(data)

        if self.head is None:
            self.head = newNode
            self.tail = newNode
            newNode.next = self.head
        else:
            self.tail.next = newNode
            self.tail = newNode
            self.tail.next = self.head

    def deleteEnd(self):
        if self.head is None:
            return

        if self.head != self.tail:
            current = self.head

            while current.next != self.tail:
                current = current.next

            self.tail = current
            self.tail.next = self.head

        else:
            self.head = None
            self.tail = None

    def display(self):
        if self.head is None:
            print("List is empty")
            return

        current = self.head

        while True:
            print(current.data, end=" ")
            current = current.next

            if current == self.head:
                break

        print("\n")


# Main
cl = CreateList()

cl.add(1)
cl.add(2)
cl.add(3)
cl.add(4)

print("Original List:")
cl.display()

while cl.head is not None:
    cl.deleteEnd()
    print("Updated List:")
    cl.display()
```
